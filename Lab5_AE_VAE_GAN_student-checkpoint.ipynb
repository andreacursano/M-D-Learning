## What do the models learn?

A digit classifier estimates something like $p(y\mid x)$: given an image, which label is most likely? A **generative model** defines a procedure for obtaining new data, such as a new image $x$. A CNN is simply a choice of layers; it can be used in a classifier, a VAE encoder, or a GAN discriminator.

An **encoder** turns an image into a latent representation; a **decoder** maps that representation back to image space. 

The **generator** in a GAN starts with random noise rather than an image. Its **discriminator** learns to distinguish real images from generated images, providing a training signal for the generator.

| Model | Training signal | Generation after training | Main difficulty |
|:--|:--|:--|:--|
| **AE** | Reconstruct its input | No reliable sampling rule | Encoded vectors need not resemble a chosen prior |
| **VAE** | Reconstruction + KL regularization | Sample $z\sim p(z)$, decode | Reconstruction and regularization compete |
| **GAN** | Real-versus-generated discriminator | Sample $z$, apply $G(z)$ | Unstable optimization and incomplete coverage |

A plain AE is primarily a **representation/reconstruction model**. The VAE and GAN are explicit generative models because they are trained with a procedure for sampling new images.

# Part I — From an autoencoder to a VAE
## 1. Compress an image and reconstruct it

The encoder $f_\phi$ maps $x\in\mathbb R^P$ to a latent vector $z\in\mathbb R^d$. The decoder $g_\theta$ maps it back to a vector of $P$ pixel values:

$$z=f_\phi(x),\qquad \hat x=\sigma(g_\theta(z)).$$

An MNIST image contains $28 \times 28 = 784$ pixels. We denote the input dimension by $P=784$. Our Autoencoder (AE) uses an **encoder** to map these 784 pixel values into a much smaller vector of $d=8$ numbers, called the **latent vector**.

The **decoder** then takes this latent vector and attempts to reconstruct the original image:

$$
x \in \mathbb{R}^{784}
\;\xrightarrow{\text{Encoder}}\;
z \in \mathbb{R}^{8}
\;\xrightarrow{\text{Decoder}}\;
\hat{x} \in \mathbb{R}^{784}.
$$

During training, the AE minimizes the difference between the original image $x$ and its reconstruction $\hat{x}$. It therefore learns to preserve information that is useful for reconstructing the training images.



*Autoencoder: the image is compressed to a latent vector and reconstructed.*

> **Task 1 — AE encoder and decoder**:
Write the fully connected encoder and decoder. The encoder maps an image of `INPUT_DIM` pixels to `LATENT_DIM` values, and the decoder maps those values back to `INPUT_DIM` image logits. Use a hidden layer of width `HIDDEN_DIM` with ReLU in each network. Do not put sigmoid in the decoder; the loss function handles it.

```python
class AutoEncoder(nn.Module):
    def __init__(self, latent_dim=LATENT_DIM):
        super().__init__()
        # Encoder dimensions: INPUT_DIM -> HIDDEN_DIM -> latent_dim.
        # The latent layer is linear; it represents a learned image code.
        self.encoder = nn.Sequential(
            nn.Linear(INPUT_DIM, HIDDEN_DIM),# YOUR CODE: Linear layer, ReLU, Linear layer.
            nn.ReLU(),
            nn.Linear(HIDDEN_DIM, latent_dim)
        )

        # Decoder dimensions: latent_dim -> HIDDEN_DIM -> INPUT_DIM.
        # Return logits; do not add sigmoid to the Sequential module.
        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, HIDDEN_DIM),# YOUR CODE: Linear layer, ReLU, Linear layer.
            nn.ReLU(),
            nn.Linear(HIDDEN_DIM, INPUT_DIM)
        )

    def forward(self, x):
        # x may be [B,P] or [B,1,S,S]. Keep the batch dimension.
        # YOUR CODE: flatten x, run the encoder, then the decoder.
        x = x.reshape(x.size(0), -1) 
        z = self.encoder(x)
        return self.decoder(z)
        
        
```

We have implemented the encoder and decoder. The next step is to **train the Autoencoder to reconstruct its input images**.

For each image, the encoder produces a latent vector, and the decoder uses this vector to reconstruct the original image. Initially, the network parameters are random, so the reconstructions will be poor. During training, we want to adjust these parameters so that the reconstructed images become increasingly similar to the originals.

#### 1. Computing the reconstruction loss

To measure how well the Autoencoder reconstructs an image, we need a **reconstruction loss** that compares the original image $x$ and the decoder output $\hat x$. We will use **binary cross-entropy (BCE)**, applied independently to each pixel of values in $[0,1]$.

For a pixel with original value $x_p$ and reconstructed value $\hat{x}_p$, the loss is

$$
\ell(x_p,\hat{x}_p)
=
-x_p\log(\hat{x}_p)
-(1-x_p)\log(1-\hat{x}_p).
$$

Here, $\hat{x}=\sigma(g_\theta)$, where $g_\theta$ is the decoder's output before the sigmoid activation (the *logit*).

The loss is minimized when the reconstructed pixel value matches the original pixel value. As the two values differ, the loss increases.

For a minibatch of $B$ images, each containing $P$ pixels, we calculate

$$
\mathcal L_{\mathrm{recon}}
=
\frac{1}{B}
\sum_{i=1}^{B}
\sum_{p=1}^{P}
\ell(x_{i,p},\hat{x}_{i,p}).
$$

Notice that we **sum the losses over all pixels in an image, then average over the images in the minibatch**.

In PyTorch, we use `F.binary_cross_entropy_with_logits`, which combines the sigmoid operation and BCE in a numerically stable way(avoids numerical issues from applying sigmoid and logarithms separately). Therefore, **do not apply sigmoid to the decoder output before calculating the loss**.

Although MNIST contains grayscale pixel values rather than strictly binary values, BCE is commonly used as a reconstruction objective for this type of data.

> **Task 2 — Reconstruction loss and optimizer update**

 > **2.1 — Implement `reconstruction_loss(logits, images)`.**

   > - The decoder output `logits` has shape `[B, P]`.
   > - The original images may have shape `[B, P]` or `[B, 1, S, S]`. Reshape them to match the decoder output.
   > - Compute BCE using `reduction="none"` to obtain individual pixel losses.
   > - Sum over the pixel dimension and average over the minibatch.
   > - Return a single scalar tensor containing the reconstruction loss.

  >  **2.2 — Updating the Autoencoder parameters: Implement `ae_train_step(model, optimizer, images)` following these steps.**

> Once we can calculate the reconstruction loss, we can use **backpropagation and gradient descent** to train the model.
    $$
        \min_{\phi,\theta}
        \mathcal L_{\mathrm{recon}}(\phi,\theta).
        $$
  We will use the Adam optimizer.

 > **Remember that a single training step consists of five operations:**
 >   1. **Clear previous gradients:** PyTorch accumulates gradients by default, so we reset them before each update.
 >   2. **Forward pass:** Pass the images through the Autoencoder to obtain reconstruction logits.
 >   3. **Compute the loss:** Compare the reconstructions with the original images using the function implemented above.
 >   4. **Backward pass:** Call `loss.backward()` to compute the gradients of the loss with respect to all trainable parameters.
 >   5. **Optimizer step:** Call `optimizer.step()` to update the encoder and decoder parameters.



```python
def reconstruction_loss(logits, images):
    """Return the pixel-summed binary cross-entropy, averaged over examples."""
    # images: [B,P] or [B,1,S,S]; logits: [B,P].
    targets = images.reshape(images.size(0), -1)# YOUR CODE: flatten images, compute BCE with reduction='none',
    bce_loss =  F.binary_cross_entropy_with_logits(logits, targets, reduction = 'none')
    return bce_loss.sum(dim = 1).mean()
    # then sum the pixel losses within each image and average over the batch.
    


def ae_train_step(model, optimizer, images):
    """Train the AE on one minibatch; return a Python float for the loss."""
    # images are already on the same device as model parameters.
    model.train()
    optimizer.zero_grad()
    logits = model(images)
    loss = reconstruction_loss(logits, images)
    loss.backward()
    optimizer.step()
    
    return loss.item()
    
```

```python
# AE training is short enough to run on a CPU; no separate training function is needed.
set_seed(SEED + 10)
ae = AutoEncoder().to(device)
optimizer = torch.optim.Adam(ae.parameters(), lr=1e-3)
ae_curve = []
for epoch in range(AE_EPOCHS):
    for images, _ in train_loader(SEED + 10 + epoch):
        images = images.to(device)  # Move each batch to the model's device.
        ae_train_step(ae, optimizer, images)

    # Validation data are never used for updating parameters.
    ae.eval()
    with torch.no_grad():
        val_images = x_val.to(device)
        val_loss = reconstruction_loss(ae(val_images), val_images).item()
    ae_curve.append(val_loss)
    print(f"AE {epoch + 1:2d}/{AE_EPOCHS}: validation reconstruction = {val_loss:.2f}")

plt.plot(range(1, AE_EPOCHS + 1), ae_curve, marker="o")
plt.xlabel("Epoch")
plt.ylabel("Validation reconstruction / image")
plt.title("AE learning curve")
plt.show()

with torch.no_grad():
    ae.eval()
    show_rows(
        {
            "Real": fixed_images,
            "AE reconstruction": ae(fixed_images.to(device)).sigmoid(),
            "AE, random z": ae.decoder(fixed_z[:8].to(device)).sigmoid(),
        },
        title="Reconstruction is not the same as generation",
    )
```

**But can we use the decoder to generate completely new images?**

One possibility would be to sample a random latent vector from a standard Gaussian distribution:

$$
z \sim \mathcal{N}(0,I)
$$

and pass it through the trained decoder.

However, **there is no guarantee that the resulting image will look like a realistic digit**.

The reason is that an ordinary AE is trained to reconstruct images using latent vectors produced by its own encoder. It is **not required to organize these vectors according to any particular probability distribution**.

For example, the encoded images might occupy a few small, irregular regions of the 8-dimensional latent space, leaving large regions with no training examples. A randomly sampled Gaussian vector may fall into one of these unfamiliar regions. Since the decoder was not trained on such vectors, its output may be unrealistic or difficult to recognize.

This reveals an important distinction:

- **Reconstruction:** The AE learns to reconstruct an existing image by encoding and decoding it.
- **Generation:** The AE does not provide a reliable way to sample new latent vectors that will produce realistic images.

**This limitation motivates Variational Autoencoders (VAEs).**

Instead of mapping each image to a single latent vector, a VAE learns a probability distribution over latent vectors. It also introduces a regularization term that encourages these distributions to remain compatible with a chosen prior, typically $\mathcal{N}(0,I)$.

This gives us a principled way to sample latent vectors and generate new images using the decoder.

## 2. The VAE: a sampling rule for new images

To generate new data, we would like to sample a latent vector from a **known prior** $p(z)=\mathcal N(0,I)$. Instead of encoding an image as one vector, the VAE encoder defines a Gaussian distribution for each image:

$$q_\phi(z\mid x)=\mathcal N\!\left(\mu_\phi(x),\operatorname{diag}(\sigma_\phi^2(x))\right).$$

The encoder predicts a **mean** $\mu$ and a **log variance** $\ell=\log\sigma^2$. A latent sample is obtained by **reparameterization**:

$$\epsilon\sim\mathcal N(0,I),\qquad z=\mu+\exp(\ell/2)\odot\epsilon.$$

This lets gradients pass from reconstruction error to the encoder because the random noise $\epsilon$ is independent of network parameters. At generation time we do not need an input image or the encoder: we sample $z\sim p(z)$ and decode it.



*VAE: training uses the encoder; generation uses the prior and decoder only.*

### Sampling and the VAE encoder

Unlike a standard Autoencoder, which maps an image directly to a latent vector $z$, a **Variational Autoencoder (VAE)** maps each input image to a probability distribution over possible latent vectors.

We assume that this distribution is Gaussian:

$$
q_\phi(z\mid x)=\mathcal N\left(\mu_\phi(x),\operatorname{diag}(\sigma_\phi^2(x))\right).
$$

Therefore, the encoder must predict two vectors:

- **Mean ($\mu$):** Determines the center of the Gaussian distribution in latent space.
- **Log variance ($\log\sigma^2$):** Determines how spread out the distribution is along each latent dimension. We predict log variance instead of variance directly because it can take any real value, while exponentiation guarantees a positive variance.

For a minibatch of $B$ images and latent dimension $d$, both vectors have shape `[B, d]`.

To reconstruct an image, we need to sample a latent vector $z$ from the distribution predicted by the encoder.

However, directly sampling from a distribution does not automatically preserve the differentiable path needed to train its parameters using ordinary backpropagation.

The **reparameterization trick** solves this by separating the random component from the learned parameters:

$$
\epsilon\sim\mathcal N(0,I)
$$

$$
\sigma=\exp\left(\frac12\log\sigma^2\right)
$$

$$
\boxed{z=\mu+\sigma\odot\epsilon}
$$

Here, $\epsilon$ is standard Gaussian noise, and $\odot$ denotes element-wise multiplication.

The resulting $z$ has the desired Gaussian distribution, but is now computed through differentiable operations involving $\mu$ and $\log\sigma^2$. This allows reconstruction-loss gradients to reach the encoder.

---

> **Task 3.1 — Implement the reparameterization trick**
>
> Implement `reparameterize(mu, logvar, eps=None)`.



> **Implementation steps:**
>
> 1. Receive `mu` and `logvar`, both with shape `[B, d]`.
> 2. Compute the standard deviation from `logvar` using `torch.exp()`.
> 3. If `eps` is `None`, generate standard Gaussian noise using `torch.randn_like()`.
> 4. Otherwise, use the given `eps`. This allows us to reuse the same random noise when comparing different models.
> 5. Compute and return $z=\mu+\sigma\odot\epsilon$.
> 6. Do not use `.detach()` or NumPy operations, since gradients must propagate through the calculation.

```python
def reparameterize(mu, logvar, eps=None):
    """Sample z = mu + std * eps while keeping gradients to the encoder."""
    # mu/logvar: [B,d]. logvar stores log(sigma^2), not log(sigma).
    std = torch.exp(0.5 * logvar)
    if eps is None :
        eps = nn.randn_like(std)
    z = mu + std * eps

    return z
```

The VAE encoder has a similar structure to the ordinary AE encoder, but instead of producing one latent vector, it produces the parameters of a Gaussian distribution.

We first extract a hidden representation:

$$
h=\operatorname{ReLU}(W_1x+b_1).
$$

From this representation, two separate linear layers predict the mean and log variance:

$$
\mu=W_\mu h+b_\mu
$$

$$
\log\sigma^2=W_\sigma h+b_\sigma.
$$

Notice that the mean and log variance are computed from the **same hidden representation**, but using different learnable parameters.

> **Task 3.2 — Implement the VAE model**
>
> Complete the `VAE` class constructor and its `encode(x)` method.

> **Implementation steps:**
>
> 1. Define `self.features`, which maps the input image from `INPUT_DIM` to `HIDDEN_DIM`, using a linear layer followed by ReLU.
> 2. Define `self.mu_head` as a linear layer mapping `HIDDEN_DIM` to `latent_dim`.
> 3. Define `self.logvar_head` with the same input and output dimensions, but separate parameters.
> 4. Define `self.decoder` using `nn.Sequential`, with the architecture `latent_dim → HIDDEN_DIM → INPUT_DIM` and a ReLU activation after the first linear layer.
> 5. In `encode(x)`, flatten the image dimensions if necessary, while preserving the batch dimension.
> 6. Pass the flattened images through `self.features`.
> 7. Pass the resulting hidden representations through both heads and return `(mu, logvar)`.

**Important:** Do not apply a sigmoid or softmax to the mean or log variance. The mean can take any real value, and the log variance is also unconstrained.


> Implement `VAE.forward(x, eps=None)`.

> **Implementation steps:**
>
> 1. Call `encode(x)` to obtain `mu` and `logvar`.
> 2. Call `reparameterize(mu, logvar, eps)` to obtain the latent sample `z`.
> 3. Pass `z` through the decoder to obtain the reconstruction logits.
> 4. Return the tuple `(logits, mu, logvar)`.

```python
class VAE(nn.Module):
    """Fully connected VAE: a shared encoder followed by two Gaussian heads."""
    def __init__(self, latent_dim=LATENT_DIM):
        super().__init__()
        self.latent_dim = latent_dim
        self.features = nn.Sequential(
            nn.Linear(INPUT_DIM, HIDDEN_DIM),
            nn.ReLU()
        )
        # YOUR CODE: two Linear heads [HIDDEN_DIM -> latent_dim].
        self.mu_head = nn.Linear(HIDDEN_DIM, latent_dim)
        self.logvar_head = nn.Linear(HIDDEN_DIM, latent_dim)
        # YOUR CODE: Sequential decoder [latent_dim -> HIDDEN_DIM -> INPUT_DIM].
        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, HIDDEN_DIM),
            nn.ReLU(),
            nn.Linear(HIDDEN_DIM, INPUT_DIM)            
        )

    def encode(self, x):
        # YOUR CODE: flatten pixels, compute shared features, return (mu, logvar).
        x_flat = x.reshape(size(0), -1)
        h = self.features(x_flat)
        mu = self.mu_head(h)
        logvar = self.logvar_head(h)
        

        return mu, logvar
        
        

    def forward(self, x, eps=None):
        mu, logvar = self.encode(x)
        z = reparametrize(mu, logvar, eps)
        
        logits = self.decoder(z)
        return logits, mu, logvar
        
```

| `logvar` | `[B, d]` | Log variance of the approximate posterior |
| `z` | `[B, d]` | Sampled latent vector |
| `logits` | `[B, P]` | Decoder output before sigmoid |

We return `mu` and `logvar` alongside the reconstruction because they will be needed to compute the **KL divergence term** in the next task.

### KL Divergence and the Complete VAE Loss

We have implemented the VAE encoder, which predicts the mean and log variance of a Gaussian distribution, and the reparameterization trick, which allows us to sample a latent vector while preserving gradient flow.

However, **simply sampling from a Gaussian distribution does not guarantee that we can generate realistic images**.

If we trained the VAE using only reconstruction loss, the encoder would be free to place the latent distributions anywhere in the latent space, just like an ordinary Autoencoder.

To make generation possible using a known prior distribution, we need to introduce an additional constraint: **the latent distributions learned by the encoder should remain compatible with the distribution from which we will sample during generation**.

This is the purpose of the **Kullback–Leibler (KL) divergence**.

#### What is KL divergence?

KL divergence measures how different two probability distributions are.

For distributions $q(z)$ and $p(z)$, it is defined as:

$$D_{\mathrm{KL}}(q\|p)=\mathbb E_{q(z)}[\log q(z)-\log p(z)].$$

KL divergence has two important properties:

- $D_{\mathrm{KL}}(q\|p)\geq 0$.
- $D_{\mathrm{KL}}(q\|p)=0$ when the two distributions are identical (almost everywhere).

Unlike a distance metric, KL divergence is **not symmetric**:

$$
D_{\mathrm{KL}}(q\|p)
\neq
D_{\mathrm{KL}}(p\|q)
$$

in general.

In our VAE, we compare two distributions:

**1. Approximate posterior — $q_\phi(z\mid x)$**

This distribution is predicted by the encoder for each input image:

$$
q_\phi(z\mid x)
=
\mathcal N
\left(
\mu_\phi(x),
\operatorname{diag}(\sigma_\phi^2(x))
\right).
$$

**2. Prior — $p(z)$**

This is the distribution from which we want to sample new latent vectors during generation.

We choose a standard Gaussian:

$$
p(z)=\mathcal N(0,I).
$$

The KL divergence is therefore

$$
D_{\mathrm{KL}}
\left(
q_\phi(z\mid x)
\|
\mathcal N(0,I)
\right).
$$

Minimizing this term encourages the encoder to produce distributions whose means remain close to zero and whose variances remain close to one.

This makes the learned latent representations more compatible with sampling from the standard Gaussian prior.

**However, minimizing KL alone would not be sufficient.** If the encoder predicted $\mu=0$ and $\sigma^2=1$ for every image, the KL would be zero, but the latent representation would contain no information about the particular input image.

Therefore, the VAE must balance two objectives:

- **Reconstruction:** Preserve enough information in $z$ to reconstruct the input image.
- **Regularization:** Keep the approximate posterior compatible with the prior distribution used for generation.

---

#### Computing KL divergence for Gaussian distributions

In our implementation, the encoder predicts

$$
\mu\in\mathbb R^d
$$

and

$$
\log\sigma^2\in\mathbb R^d.
$$

Because we assume a diagonal Gaussian approximate posterior and a standard Gaussian prior, the KL divergence has a closed-form expression:

$$
\boxed{
D_{\mathrm{KL}}
=
\frac12
\sum_{j=1}^{d}
\left(
\mu_j^2+
\sigma_j^2-
1-
\log\sigma_j^2
\right)
}
$$

Since our encoder predicts `logvar` instead of variance, we use

$$
\sigma_j^2=\exp(\text{logvar}_j).
$$

Therefore, the formula becomes

$$
\boxed{
D_{\mathrm{KL}}
=
-\frac12
\sum_{j=1}^{d}
\left(
1+\text{logvar}_j-\mu_j^2
-\exp(\text{logvar}_j)
\right)
}
$$

**Example**

Suppose the encoder predicts

$$
\mu=[0,0,0],
\qquad
\log\sigma^2=[0,0,0].
$$

Then the approximate posterior is exactly $\mathcal N(0,I)$, and

$$
D_{\mathrm{KL}}=0.
$$

Now suppose

$$
\mu=[2,2,2],
\qquad
\log\sigma^2=[0,0,0].
$$

Each coordinate contributes

$$
\frac12(2^2+1-1-0)=2.
$$

Since there are three coordinates, the total KL is

$$
D_{\mathrm{KL}}=6.
$$

This illustrates how the KL term penalizes latent distributions whose means are far from the prior's mean.

---

#### The complete VAE objective

We can now combine reconstruction loss and KL divergence.

Recall that the reconstruction term encourages the decoder to reproduce the input image:

$$
\mathcal L_{\mathrm{recon}}
=
\frac{1}{B}
\sum_{i=1}^{B}
\mathcal L_{\mathrm{BCE}}(x_i,\hat{x}_i),
$$

where the pixelwise BCE terms are summed within each image.

The KL term encourages the approximate posterior to remain close to the prior:

$$
\mathcal L_{\mathrm{KL}}
=
\frac1B
\sum_{i=1}^{B}
D_{\mathrm{KL}}
\left(
q_\phi(z\mid x_i)
\|
p(z)
\right).
$$

The complete VAE objective is

$$
\boxed{
\mathcal L_{\mathrm{VAE}}
=
\mathcal L_{\mathrm{recon}}
+
\beta\mathcal L_{\mathrm{KL}}
}
$$

where $\beta$ controls the strength of the KL regularization.

For $\beta=1$, this corresponds to the standard VAE objective, commonly derived by maximizing the **Evidence Lower Bound (ELBO)**:

$$
\operatorname{ELBO}(x)
=
\mathbb E_{q_\phi(z\mid x)}
[\log p_\theta(x\mid z)]
-
D_{\mathrm{KL}}
\left(q_\phi(z\mid x)\|p(z)\right).
$$

Maximizing the ELBO is equivalent to minimizing its negative, consisting of a reconstruction term and KL divergence. Our pixelwise BCE provides the reconstruction objective used in this lab.

The parameter $\beta$ allows us to modify the balance between these two terms:

- **$\beta=0$:** The model is trained only to reconstruct images. There is no pressure to match the prior.
- **$\beta=1$:** The standard VAE objective balances reconstruction and KL regularization.
- **$\beta>1$:** The KL term receives more weight, encouraging stronger prior matching, potentially at the expense of reconstruction quality.

If the KL regularization becomes too strong, the encoder may learn to produce nearly identical latent distributions for all inputs. The decoder may then ignore the latent vectors, a phenomenon known as **posterior collapse**.

This trade-off motivates the KL annealing experiment in the next section.

# Part II — Generative Adversarial Networks
## 3. Generate images without an encoder — Learn a generator from a real-versus-generated decision

The VAE trains its decoder by reconstructing a particular input image. A **generative adversarial network (GAN)** takes a different approach: it learns to produce realistic images **without pairing each random latent vector with a target image**.

A GAN contains two neural networks:

- The **generator** $G_\theta$ receives random noise $z\sim\mathcal N(0,I)$ and produces an image $\tilde x=G_\theta(z)$. It learns how to map samples from a simple distribution into image space.
- The **discriminator** $D_\psi$ receives either a real training image $x$ or a generated image $\tilde x$ and predicts whether the image is real. It distinguishes **real/fake**, not the identity of the digit.

The discriminator returns one **logit** $a_\psi(x)\in\mathbb R$ for each image. Applying sigmoid converts this to a probability:

$$
D_\psi(x)=\sigma(a_\psi(x))=\frac{1}{1+e^{-a_\psi(x)}}.
$$

A score near 1 means that the current discriminator believes the image is real; a score near 0 means that it believes the image was generated. **In our PyTorch code, the discriminator returns $a_\psi(x)$, not $D_\psi(x)$**. We pass the logits directly to `binary_cross_entropy_with_logits` for numerical stability.

The noise distribution and generator jointly define a distribution of images $p_G$. We can sample from it simply by drawing $z$ and computing $G_\theta(z)$, even though we cannot usually evaluate the exact density $p_G(x)$. This is why a GAN is called an **implicit generative model**. Once training is complete, the generator can produce new images by itself; the discriminator is only needed during training.

## GAN loss functions: what does each network learn?

The two networks have **different objectives**. The discriminator learns a binary classification task, while the generator learns to make the discriminator classify its generated images as real.

### Discriminator loss

For a minibatch of real images $x$ and generated images $G(z)$, the discriminator should assign probability 1 to real images and 0 to generated ones. The loss we **minimize** is

$$
\boxed{
\mathcal L_D
=\tfrac12\left(
-\mathbb E_{x\sim p_{\mathrm{data}}}[\log D_\psi(x)]
-\mathbb E_{z\sim p(z)}[\log(1-D_\psi(G_\theta(z)))]
\right).
}
$$

Each expectation is estimated by a minibatch mean. The factor $\tfrac12$ gives equal weight to the two terms and matches our implementation: `0.5 * (real_loss + fake_loss)`. It is only a scaling convention; some formulations use the sum without this factor.

### Generator loss

The original GAN formulation is a two-player minimax game:

$$
\min_G\max_D
\mathbb E_{x\sim p_{\mathrm{data}}}\log D(x)
+\mathbb E_{z\sim p(z)}\log(1-D(G(z))).
$$

For a fixed discriminator, the original **minimax** generator loss is

$$
\mathcal L_G^{\mathrm{MM}}=\mathbb E_z[\log(1-D(G(z)))].
$$

Although this has the right objective, it can provide a weak gradient when the discriminator confidently rejects generated images. We therefore use the **non-saturating generator loss**:

$$
\boxed{\mathcal L_G^{\mathrm{NS}}=-\mathbb E_z[\log D(G(z))].}
$$

This asks the generator to produce images that the discriminator would classify as real. In code, it is `binary_cross_entropy_with_logits(fake_logits, ones)`. **This does not change the discriminator's real/fake targets**: we use this target only while optimizing the generator.

### Why use the non-saturating loss?

Consider one generated image, and let $a$ be its discriminator logit and $s=\sigma(a)$. Differentiating the two generator losses gives

$$
\frac{\partial\log(1-s)}{\partial a}=-s,
\qquad
\frac{\partial(-\log s)}{\partial a}=-(1-s).
$$

If $D(G(z))=s=0.01$, the magnitudes of these derivatives are **0.01** for minimax and **0.99** for non-saturating loss. Both encourage an increase in the discriminator's score for generated images, but the non-saturating loss supplies a much stronger *local logit gradient* in this situation.

This does **not** mean that the generator parameter update is 99 times larger: gradients must also pass through the discriminator and generator, and the optimizer transforms them.

In summary, during the initial iterations of GAN training, using non-saturation loss for the generator helps maintain sufficient gradient magnitudes even when the discriminator is strong. This approach mitigates the vanishing gradient problem, facilitating more effective learning and convergence in GANs.

### What is the GAN learning? (An idealized argument)

For a fixed generator, the optimal discriminator in the original minimax game is

$$
D^*(x)=\frac{p_{\mathrm{data}}(x)}{p_{\mathrm{data}}(x)+p_G(x)}.
$$

When the generated and real distributions match, this optimal discriminator assigns probability $1/2$ wherever they have positive density. This helps explain why an adversarial classifier can provide a signal for learning a distribution. It is an **idealized result**, not a guarantee that finite neural networks will converge to the correct distribution.

## Alternating updates and gradient flow

GAN training is not one ordinary supervised optimization step. **The networks take turns**, with separate optimizers and separate losses.

**Step A — Train the discriminator (keep the generator fixed).**

1. Draw a minibatch of real images $x$ and a batch of Gaussian noise $z$.
2. Produce generated images $\tilde x=G(z)$.
3. Compute $D(x)$ and $D(\tilde x)$, then minimize $\mathcal L_D$.
4. Call `opt_D.step()` to change only discriminator parameters.

Use `fake = G(z).detach()` for this step. Detaching the generated image prevents the discriminator's loss from propagating gradients back into the generator. The generated images are simply training examples for $D$ in this update.

**Step B — Train the generator (keep the discriminator fixed).**

1. Draw a new batch of noise $z$ and compute $\tilde x=G(z)$.
2. Pass those generated images through the current discriminator and compute $\mathcal L_G^{\mathrm{NS}}$.
3. Backpropagate along the entire path $z\rightarrow G(z)\rightarrow D(G(z))\rightarrow\mathcal L_G$.
4. Call `opt_G.step()` to change only generator parameters.

The discriminator's **weights must remain unchanged** in Step B, but its forward operations must remain differentiable with respect to the generated image. Setting `requires_grad_(False)` on $D$'s parameters allows this; calling `.detach()` on $G(z)$ or enclosing $D(G(z))$ in `torch.no_grad()` would break the required gradient path.

**Remember:** `loss.backward()` computes gradients; `optimizer.step()` changes parameters. Computing a gradient *through* the discriminator does not mean we should update its weights during the generator step.

| Update | Network whose parameters change | Should `G(z)` be detached? |
|---|---|---|
| Discriminator | $D$ only | Yes |
| Generator | $G$ only | No |

### GAN limitations: images, losses, and diversity

The GAN learns a distribution of images rather than one-to-one reconstructions. A basic GAN therefore has **no encoder** for a real image and cannot directly reconstruct that image.

GAN training can also be unstable: while $G$ improves, $D$ changes too, so each network faces a changing optimization problem. In **mode collapse**, the generator produces only a small subset of the data distribution—for example, recognizable 7s but almost no other digits, or many nearly identical 7s. A low generator loss does not prove good visual quality, and an average discriminator score near $0.5$ does not prove successful distribution matching: the discriminator might simply be weak.

This is why we finish by comparing **generated images and digit variety**, not only the two GAN losses.

```python
def one_hot(labels):
    # One-hot vectors are float tensors because they are concatenated with features.
    return F.one_hot(labels.long(), num_classes=10).float()


class ConditionalVAE(nn.Module):
    """Conditional VAE with a Gaussian, Laplace, or learned Gaussian prior."""
    def __init__(self, kind="gaussian"):
        super().__init__()
        assert kind in {"gaussian", "laplace", "learned"}
        self.kind = kind
        self.posterior = nn.Sequential(
            nn.Linear(INPUT_DIM + 10, HIDDEN_DIM),
            nn.ReLU(),
            nn.Linear(HIDDEN_DIM, 2 * LATENT_DIM),
        )
        self.decoder = nn.Sequential(
            nn.Linear(LATENT_DIM + 10, HIDDEN_DIM),
            nn.ReLU(),
            nn.Linear(HIDDEN_DIM, INPUT_DIM),
        )
        # Only the learned prior has label-dependent parameters.
        self.prior_network = (
            nn.Sequential(nn.Linear(10, 32), nn.ReLU(), nn.Linear(32, 2 * LATENT_DIM))
            if kind == "learned" else None
        )

    def encode(self, x, y):
        features = torch.cat([x.reshape(len(x), -1), one_hot(y)], dim=1)
        mu, logvar = self.posterior(features).chunk(2, dim=1)
        return mu, logvar

    def prior_params(self, y):
        if self.prior_network is None:
            # torch.zeros(..., device=y.device) also works when a GPU is available.
            zeros = torch.zeros(len(y), LATENT_DIM, device=y.device)
            return zeros, zeros
        mu, logvar = self.prior_network(one_hot(y)).chunk(2, dim=1)
        return mu, logvar.clamp(-6, 6)

    def forward(self, x, y):
        mu, logvar = self.encode(x, y)
        z = reparameterize(mu, logvar)
        logits = self.decoder(torch.cat([z, one_hot(y)], dim=1))
        return logits, mu, logvar, z
```
